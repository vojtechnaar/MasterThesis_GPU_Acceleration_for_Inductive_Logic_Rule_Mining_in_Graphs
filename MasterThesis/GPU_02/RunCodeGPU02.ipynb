## Profiling (Nsight Systems)

Recompiles with `-lineinfo` for better GPU kernel attribution, then runs **Nsight Systems** (`nsys`) to capture a full timeline of CUDA API calls and kernel execution.

```python
## Step 1: Recompile with -lineinfo so nsys can map samples back to source lines
!nvcc -O3 -lineinfo -std=c++17 bench_gpu.cu FinalRule.cpp RdfIndexes.cpp RuleParser.cpp \
    $(pkg-config --cflags --libs raptor2) -o bench_gpu_profiled
```

```python
## Step 2: Run Nsight Systems profile (100-rule file for speed)
## Captures CUDA API, kernel launches, and memory transfers
!nsys profile \
    --stats=true \
    --output=profile_gpu02 \
    --force-overwrite=true \
    ./bench_gpu_profiled \
        ../test_data/original_train.ttl \
        ../test_data/rules_150minutes_short_100rules.txt \
    2>&1 || echo "nsys not found, falling back to nvprof" && \
nvprof --print-gpu-summary \
    ./bench_gpu_profiled \
        ../test_data/original_train.ttl \
        ../test_data/rules_150minutes_short_100rules.txt \
    2>&1
```

```python
## Step 3: Print Nsight Systems kernel-level statistics from the saved report
## Shows GPU kernel durations, CUDA API times, and memory transfer breakdown
!nsys stats profile_gpu02.nsys-rep 2>&1 || echo "(nsys stats unavailable; see nsys output above)"
```
